#Build & Run Python App in Docker – Step-by-Step

#✅ Step 1: Create Your Python App
Make a folder and create a simple Python app.

Folder Structure:

python-docker-app/

├── app.py

├── requirements.txt

├── Dockerfile


🧱 1. Create a Project Directory

CMD >> mkdir python-docker-app

CMD >> cd python-docker-app






```python
#app.py

from flask import Flask

app = Flask(__name__)

@app.route('/')
def hello():
    return "Hello from Dockerized Python App!"

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000)
```

```python
#requirements.txt

CMD >> echo flask > requirements.txt

#This tells Docker to install Flask.
```

#✅ Step 2: Create a Dockerfile
This file tells Docker how to build your app image.



```python
#📌 Linux / macOS:

CMD >> touch Dockerfile


#📌 Windows CMD:

CMD >> echo.> Dockerfile
```

```python
##Dockerfile


# Use official Python base image
FROM python:3.10-slim

# Set working directory
WORKDIR /app

# Copy code to container
COPY . .

# Install dependencies
RUN pip install --no-cache-dir -r requirements.txt

# Expose port
EXPOSE 5000

# Run the app
CMD ["python", "app.py"]
```

#✅ Step 3: Build the Docker Image
Open terminal in the python-docker-app folder:

CMD >> docker build -t python-docker-app .

-t tags the image for easier reference.

#✅ Step 4: Run the Container

CMD >> docker run -p 5000:5000 python-docker-app

This maps container’s port 5000 to your localhost 5000.

#✅ Step 5: Open the App

Visit http://localhost:5000
You should see:

OUTPUT > Hello from Dockerized Python App!
