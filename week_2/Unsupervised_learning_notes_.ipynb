# Dimensionality Reduction using Principal Component Analysis (PCA)

#📌 1. What is Dimensionality Reduction?

The process of reducing the number of input variables (features) in a dataset.

Helps in:

>Reducing overfitting

>Speeding up algorithms

>Visualizing data

>Removing noise

#📌 2. Introduction to PCA

#🎯 Principal Component Analysis (PCA)

>PCA is an unsupervised linear transformation technique used for dimensionality reduction.

>It transforms the original features into a new set of orthogonal components (principal components) that capture the maximum variance in the data.

>These components are ranked — the first few retain most of the variance.

#📌 3. Key Concepts Behind PCA

| Term                     | Explanation                                                                      |
| ------------------------ | -------------------------------------------------------------------------------- |
| **Principal Components** | New axes formed by linear combinations of the original features                  |
| **Variance**             | Measure of data spread; PCA tries to retain directions with the highest variance |
| **Orthogonal Vectors**   | Principal components are uncorrelated and form orthogonal axes                   |
| **Eigenvectors**         | Directions of the new feature space                                              |
| **Eigenvalues**          | Measure the magnitude of variance captured by each eigenvector                   |

#📘 K-Means Clustering

#📌 1. What is Clustering?

Clustering is an unsupervised machine learning technique used to group similar data points into clusters.

The algorithm tries to maximize intra-cluster similarity and minimize inter-cluster similarity.

#📌 2. Introduction to K-Means

🎯 K-Means Clustering

>K-Means partitions n data points into k distinct, non-overlapping clusters.

>Each cluster is represented by the centroid (mean) of its points.

>The objective is to minimize the Within-Cluster Sum of Squares (WCSS).

#📌 3. Intuition Behind K-Means

>Place k centroids randomly.

>Assign each point to the nearest centroid.

>Recalculate the centroids as the mean of assigned points.

>Repeat until convergence (no change in assignments or centroids).

#📌 4. Objective Function (Loss Function)

🔢 Minimize the Sum of Squared Distances

#📌 5. K-Means Algorithm Steps

#🧮 Step-by-Step

1.Choose number of clusters k

2.Initialize k centroids randomly

3.Repeat until convergence:

>Assign each data point to the nearest centroid (based on distance).

>Update each centroid to be the mean of the points assigned to it.

#📌 6. Convergence Criteria

>Centroids do not change between iterations.

>Cluster assignments are stable.

>Typically achieved in a small number of iterations.

#📌 7. Choosing the Value of k

#📊 Elbow Method

Plot WCSS vs. number of clusters.

The elbow point (where WCSS starts to decrease slowly) is the optimal k.

#📌 8. Distance Metrics

By default, K-Means uses Euclidean distance:

Note: K-Means performs poorly with non-Euclidean distances or categorical data.

# 9. Strengths of K-Means

✅ Easy to implement

✅ Scales well with large datasets

✅ Fast and efficient

✅ Works well with spherical and equally sized clusters

#📌 10. Limitations of K-Means

❌ Need to specify k in advance

❌ Sensitive to initial centroid positions

❌ Can get stuck in local minima

❌ Assumes spherical, equal-size clusters

❌ Poor performance on non-convex clusters

#📊 Silhouette Score


The Silhouette Score measures how similar a data point is to its own cluster (cohesion) compared to other clusters (separation).

It helps evaluate the quality of clustering without needing ground truth labels.

>Values range from -1 to +1:

>+1: Perfectly matched to its own cluster, and far from others

>0: On or near the decision boundary between clusters

>–1: Wrongly assigned to the cluster

#Interpretation of Values

| Silhouette Score $s(i)$ | Interpretation                   |
| ----------------------- | -------------------------------- |
| Near +1                 | Well-clustered                   |
| Around 0                | On the boundary between clusters |
| Near -1                 | Incorrectly clustered            |


#Average Silhouette Score

To evaluate the whole clustering model, compute the mean silhouette score for all samples.

image.png

Higher value → Better-defined clusters.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Generate sample data
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.6, random_state=0)

# K-Means Clustering
kmeans = KMeans(n_clusters=4, random_state=0)
labels = kmeans.fit_predict(X)

# Silhouette Score
score = silhouette_score(X, labels)
print("Silhouette Score:", score)
```

#When to Use Silhouette Score

To compare clustering results for different values of k (e.g., in K-Means).

To assess how natural the clustering is.

Works well with distance-based clustering algorithms.

```python
#Visualizing Silhouette Score

from sklearn.metrics import silhouette_samples
import matplotlib.pyplot as plt
import numpy as np

sil_samples = silhouette_samples(X, labels)
cluster_labels = np.unique(labels)

for i in cluster_labels:
    ith_silhouette_vals = sil_samples[labels == i]
    ith_silhouette_vals.sort()
    plt.barh(range(len(ith_silhouette_vals)), ith_silhouette_vals)

plt.axvline(x=score, color='red', linestyle='--')
plt.xlabel("Silhouette Coefficient")
plt.ylabel("Cluster")
plt.title("Silhouette Plot")
plt.show()
```

#Summary

The Silhouette Score is a powerful tool to measure clustering quality.

Helps determine the optimal number of clusters (k).

Should be used in conjunction with other methods like the Elbow Method, especially in K-Means clustering.
