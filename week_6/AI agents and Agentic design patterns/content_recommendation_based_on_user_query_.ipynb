```python
#1️⃣ agent_graph.py – LangGraph + Gemini Agent Logic

%%writefile agent_graph.py


import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import Tool, create_react_agent, AgentExecutor
from langchain import hub
from langgraph.graph import StateGraph, END
from typing import TypedDict

# Load API Key
from dotenv import load_dotenv
load_dotenv()

# Tool function
def recommend_books_and_movies(user_interest: str) -> str:
    recommendations = {
        "ai": "Book: 'Life 3.0', Movie: 'Ex Machina'",
        "romance": "Book: 'Pride and Prejudice', Movie: 'The Notebook'",
        "sci-fi": "Book: 'Dune', Movie: 'Interstellar'",
        "history": "Book: 'Sapiens', Movie: 'The King's Speech'",
    }
    for key in recommendations:
        if key in user_interest.lower():
            return f"Recommended for {key}: {recommendations[key]}"
    return "Try being more specific like 'romance', 'sci-fi', or 'history'."

recommendation_tool = Tool(
    name="recommend_books_and_movies",
    func=recommend_books_and_movies,
    description="Recommends books and movies based on the user's interest.",
)

llm = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",
    temperature=0.3,
    convert_system_message_to_human=True
)

prompt = hub.pull("hwchase17/react")
agent = create_react_agent(llm=llm, tools=[recommendation_tool], prompt=prompt)
agent_executor = AgentExecutor(agent=agent, tools=[recommendation_tool], verbose=False)

# LangGraph Setup
class AgentState(TypedDict):
    user_input: str
    result: str

def run_agent(state: AgentState) -> AgentState:
    response = agent_executor.invoke({"input": state["user_input"]})
    return {"user_input": state["user_input"], "result": response["output"]}

# Build LangGraph
def build_agent_graph():
    workflow = StateGraph(AgentState)
    workflow.add_node("ContentRecommender", run_agent)
    workflow.set_entry_point("ContentRecommender")
    workflow.set_finish_point("ContentRecommender")
    return workflow.compile()
```

```python
#2️⃣ app.py – Streamlit App + Ngrok Tunnel

%%writefile app.py

# app.py

import streamlit as st
from agent_graph import build_agent_graph
from pyngrok import ngrok

st.set_page_config(page_title="Content Recommender Agent", layout="centered")

st.title("📚🎬 AI Content Recommender")
st.markdown("Enter your interests (e.g., 'AI', 'romance', 'history') to get book and movie suggestions.")

graph_executor = build_agent_graph()

user_input = st.text_input("What are you interested in?")

if st.button("Get Recommendations") and user_input:
    with st.spinner("Thinking..."):
        result = graph_executor.invoke({"user_input": user_input})
        st.success(result["result"])

# Ngrok tunnel display
if "public_url" not in st.session_state:
    public_url = ngrok.connect(8501)
    st.session_state["public_url"] = public_url
    st.markdown(f"🌐 Public URL: [Open in Browser]({public_url})")
```
