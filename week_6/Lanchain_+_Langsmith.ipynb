#2) Import libraries

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

#3) Create a Gemini LLM object

```python
# Gemini 1.5 Flash model
llm = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",
    temperature=0.3,
    max_output_tokens=512,
)
```

#4) Create a simple prompt

We’ll use LangChain LCEL to wire:

User Input → Prompt → Gemini → String Parser.

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant who answers in a clear and concise way."),
    ("human", "{question}")
])

chain = prompt | llm | StrOutputParser()
```

#5) Run in a notebook cell

```python
question = "Explain the concept of Agentic AI in two sentences."
answer = chain.invoke({"question": question})
print(answer)
```

#6) Test with multiple questions in a loop

```python
questions = [
    "What is LangSmith and why is it used?",
    "Give 3 use cases of Gemini 1.5 Flash.",
    "Explain prompt engineering in simple terms."
]

for q in questions:
    print(f"Q: {q}")
    print("A:", chain.invoke({"question": q}))
    print("-" * 80)
```
