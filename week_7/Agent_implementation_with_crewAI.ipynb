# CrewAI Lab - Agent Implementation with CrewAI

### Lab Objective
Learn how to implement individual agents and multi-agent teams using CrewAI framework for collaborative AI workflows.

## What is CrewAI?
CrewAI is a revolutionary framework that brings the concept of "AI teams" to life. Think of it as creating a virtual company where each AI agent has a specific role, expertise, and responsibilities, and they collaborate just like human teams do.
Key CrewAI Principles:

* Role-Based Agents: Each agent has a specific job function and expertise
* Task-Oriented Workflows: Agents work on defined tasks with clear deliverables
* Sequential & Parallel Processing: Teams can work in order or simultaneously
* Quality Control: Built-in review and improvement mechanisms
* Scalable Architecture: From simple single-agent tasks to complex multi-agent organizations

## Use Case: Automated blog creation

* Research Agent: Gathers current information and trends
* Writer Agent: Creates engaging, SEO-optimized content
* Editor Agent: Reviews for grammar, clarity, and brand voice
* Designer Agent: Creates visuals and formatting
* Publisher Agent: Schedules and distributes content

### Sigle-Agent Blog Creation Team:

```python
import os
import json
from crewai import Agent, Task, Crew, Process
from google.colab import userdata

# Retrieve the JSON string from the secret and write it to a file
service_account_json = userdata.get('GOOGLE_SERVICE_ACCOUNT_JSON')
with open('/content/service-account-key.json', 'w') as f:
    json.dump(json.loads(service_account_json), f)
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = '/content/service-account-key.json'

# Optional: Set GOOGLE_API_KEY if required by other services (e.g., Google AI Studio)
os.environ['GOOGLE_API_KEY'] = userdata.get('GOOGLE_API_KEY')

# Create a research agent using Vertex AI
researcher = Agent(
    role='Research Specialist',
    goal='Find and analyze comprehensive information on given topics',
    backstory="""You are an expert researcher with years of experience in gathering
    and analyzing information from various sources. You have a keen eye for detail
    and can quickly identify key insights and trends. You always provide well-structured,
    accurate, and actionable information.""",
    verbose=True,
    allow_delegation=False,
    llm="vertex_ai/gemini-2.0-flash"
)

# Create a research task
research_task = Task(
    description="""Research the top 5 benefits of artificial intelligence in education.
    For each benefit:
    1. Provide a clear title
    2. Give a detailed explanation (2-3 sentences)
    3. Include a real-world example or use case
    4. Mention any potential challenges or considerations

    Make sure your research is current and focuses on practical applications.""",
    expected_output="""A well-structured report with 5 AI benefits in education,
    each containing: title, detailed explanation, real-world example, and considerations.""",
    agent=researcher
)

# Create crew
research_crew = Crew(
    agents=[researcher],
    tasks=[research_task],
    verbose=True,
    process=Process.sequential
)

# Execute the crew
result = research_crew.kickoff()
print(result)
```

### Multi-Agent Blog Creation Team:

```python
from crewai import Agent, Task, Crew, Process

# Research Agent
research_agent = Agent(
    role='Content Researcher',
    goal='Gather comprehensive information and current trends on given topics',
    backstory="""You are a meticulous researcher who excels at finding the most
    current and relevant information. You have access to various sources and can
    quickly identify trending topics and key insights.""",
    verbose=True,
    allow_delegation=False,
    llm="vertex_ai/gemini-2.0-flash"
)

# Writer Agent
writer_agent = Agent(
    role='Content Writer',
    goal='Create engaging, well-structured, and SEO-optimized blog content',
    backstory="""You are a skilled content writer with expertise in creating
    engaging blog posts. You understand SEO best practices and know how to
    structure content that both readers and search engines love.""",
    verbose=True,
    allow_delegation=False,
    llm="vertex_ai/gemini-2.0-flash"
)

# Editor Agent
editor_agent = Agent(
    role='Content Editor',
    goal='Review and improve content for clarity, grammar, and brand consistency',
    backstory="""You are an experienced editor with a sharp eye for detail.
    You ensure all content meets high standards for grammar, clarity, tone,
    and brand consistency before publication.""",
    verbose=True,
    allow_delegation=False,
    llm="vertex_ai/gemini-2.0-flash"
)

# Define tasks
research_task = Task(
    description="""Research the latest trends in remote work technology for 2025.
    Focus on tools, platforms, and innovations that are gaining popularity.""",
    expected_output="A comprehensive research report with key findings and trends",
    agent=research_agent
)

writing_task = Task(
    description="""Using the research findings, write a 800-word blog post titled
    "Top Remote Work Tech Trends to Watch in 2025". Include an engaging introduction,
    main sections covering key trends, and a compelling conclusion.""",
    expected_output="A complete, engaging blog post ready for review",
    agent=writer_agent
)

editing_task = Task(
    description="""Review the blog post for grammar, clarity, flow, and SEO optimization.
    Ensure it has proper headings, is easy to read, and maintains a professional yet
    approachable tone.""",
    expected_output="A polished, publication-ready blog post",
    agent=editor_agent
)

# Create the blog creation crew
blog_crew = Crew(
    agents=[research_agent, writer_agent, editor_agent],
    tasks=[research_task, writing_task, editing_task],
    verbose=True,
    process=Process.sequential
)

# Execute the blog creation process
blog_result = blog_crew.kickoff()
print(blog_result)
```
