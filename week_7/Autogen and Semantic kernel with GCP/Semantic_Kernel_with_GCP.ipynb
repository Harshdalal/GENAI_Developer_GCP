## Cell 4: Define Enhanced GCP Agent

```python
class GeminiChatCompletion:
    """Custom Gemini Chat Completion for Semantic Kernel"""

    def __init__(self, api_key, model_name="gemini-1.5-flash"):
        # Configure Gemini API with provided API key
        genai.configure(api_key=api_key)
        # Initialize Gemini Generative Model
        self.model = genai.GenerativeModel(model_name)
        # Service ID to register inside Semantic Kernel
        self.service_id = "gemini"

    async def complete_chat_async(self, messages, **kwargs):
        """Complete chat using Gemini"""
        try:
            # Convert incoming messages (system, user, assistant) into a single prompt
            prompt_parts = []
            for message in messages:
                # Handle Semantic Kernel message object
                if hasattr(message, 'role') and hasattr(message, 'content'):
                    role = message.role
                    content = message.content
                # Handle plain dict messages
                elif isinstance(message, dict):
                    role = message.get('role', '')
                    content = message.get('content', '')
                # Handle raw text (fallback)
                else:
                    content = str(message)
                    role = 'user'

                # Format based on role type
                if role == 'system':
                    prompt_parts.append(f"System: {content}")
                elif role == 'user':
                    prompt_parts.append(f"User: {content}")
                elif role == 'assistant':
                    prompt_parts.append(f"Assistant: {content}")
                else:
                    prompt_parts.append(content)

            # Join all parts into a single string prompt
            full_prompt = "\n".join(prompt_parts)

            # Call Gemini API to generate response
            response = self.model.generate_content(full_prompt)

            # Wrap response in a ChatMessage object (to match Semantic Kernel format)
            class ChatMessage:
                def __init__(self, content):
                    self.content = content
                    self.role = "assistant"

            return ChatMessage(response.text)

        except Exception as e:
            # Handle errors gracefully
            print(f"Gemini completion error: {str(e)}")

            # Return error as assistant message (so kernel flow doesn't break)
            class ErrorMessage:
                def __init__(self, error):
                    self.content = f"Error: {str(error)}"
                    self.role = "assistant"
            return ErrorMessage(e)


# -------------------------
# Integrating Gemini with Semantic Kernel
# -------------------------

# Create a new Semantic Kernel instance
kernel = sk.Kernel()

# Initialize Gemini service with API key
gemini_service = GeminiChatCompletion(GEMINI_API_KEY)

# Register Gemini service inside the kernel
kernel.add_service(gemini_service)

print("✅ Kernel ready with Gemini 1.5 Flash")
```

```python
class EnhancedGCPAgent:
    """Enhanced GCP Agent with better file handling"""

    def __init__(self):
        # Initialize GCP Storage client with your project
        self.client = storage.Client(project="My First Project")  # Replace with your GCP project name
        # Set the target bucket name
        self.bucket_name = "bucket_demo79"  # Replace with your bucket name
        # Define local working directory (default: Google Colab's /content/)
        self.base_dir = Path("/content/")  # Local directory for file handling

    @kernel_function(description="List files in bucket and local directory")
    def list_files(self) -> str:
        """List files available in both the GCP bucket and local directory"""
        try:
            # Fetch all files from GCP bucket
            bucket = self.client.bucket(self.bucket_name)
            bucket_files = [blob.name for blob in bucket.list_blobs()]

            # Fetch all files from local directory
            local_files = []
            if self.base_dir.exists():
                local_files = [f.name for f in self.base_dir.iterdir() if f.is_file()]

            # Return combined result
            return f"Bucket files: {bucket_files}\nLocal files: {local_files}"
        except Exception as e:
            return f"Error listing files: {str(e)}"

    @kernel_function(description="Upload file to bucket with smart filename matching")
    def upload_file(self, filename: str) -> str:
        """Upload file to bucket with smart filename matching for common variations"""
        try:
            # Try common filename variations (handles naming mismatches)
            possible_names = [
                filename,
                filename.replace('_', ' '),
                filename.replace(' ', '_'),
                filename.replace('eer2', 'eer_2'),
                filename.replace('eer_2', 'eer2')
            ]

            file_path = None
            actual_filename = None

            # Find matching file locally
            for name in possible_names:
                test_path = self.base_dir / name
                if test_path.exists():
                    file_path = test_path
                    actual_filename = name
                    break

            # If no matching file found, return available options
            if file_path is None:
                if self.base_dir.exists():
                    available_files = [f.name for f in self.base_dir.iterdir() if f.is_file()]
                    return f"❌ File '{filename}' not found. Available files: {available_files}"
                else:
                    return f"❌ Directory '{self.base_dir}' does not exist"

            # Upload file to GCP bucket
            bucket = self.client.bucket(self.bucket_name)
            blob = bucket.blob(filename)  # Use requested filename for destination in bucket
            blob.upload_from_filename(str(file_path))

            return f"✅ Uploaded '{actual_filename}' as '{filename}' to {self.bucket_name}"

        except Exception as e:
            return f"❌ Upload error: {str(e)}"

    @kernel_function(description="Get detailed file info")
    def get_file_info(self, filename: str) -> str:
        """Fetch metadata (size, creation time, type) for a file stored in the bucket"""
        try:
            bucket = self.client.bucket(self.bucket_name)
            blob = bucket.blob(filename)
            if blob.exists():
                # Reload ensures latest metadata is fetched
                blob.reload()
                return (
                    f"File: {filename}\n"
                    f"Size: {blob.size} bytes\n"
                    f"Created: {blob.time_created}\n"
                    f"Type: {blob.content_type}"
                )
            return f"File {filename} not found in bucket"
        except Exception as e:
            return f"Error getting file info: {str(e)}"

    @kernel_function(description="Check local files in directory")
    def check_local_files(self) -> str:
        """List files present locally with file sizes"""
        try:
            # Ensure directory exists
            if not self.base_dir.exists():
                return f"❌ Directory {self.base_dir} does not exist"

            files = []
            # Collect filenames with sizes
            for file_path in self.base_dir.iterdir():
                if file_path.is_file():
                    size = file_path.stat().st_size
                    files.append(f"{file_path.name} ({size} bytes)")

            # Return results
            if files:
                return f"Local files in {self.base_dir}:\n" + "\n".join(files)
            else:
                return f"No files found in {self.base_dir}"
        except Exception as e:
            return f"Error checking local files: {str(e)}"


# -------------------------
# Register Enhanced GCP Agent in Semantic Kernel
# -------------------------

# Create instance of GCP Agent
gcp_agent = EnhancedGCPAgent()

# Add it as a plugin inside Semantic Kernel (accessible as "gcp")
kernel.add_plugin(gcp_agent, plugin_name="gcp")

print("✅ Enhanced GCP Agent added to Semantic Kernel")
```

## Cell 5: Define Analysis Agent

```python
class AnalysisAgent:
    """Analysis Agent using Gemini 1.5 Flash"""

    def __init__(self):
        # Configure Gemini API with your API key
        genai.configure(api_key=GEMINI_API_KEY)
        # Initialize Gemini Generative Model (using Flash variant for fast responses)
        self.model = genai.GenerativeModel('gemini-1.5-flash')

    def analyze_files(self, file_data):
        """
        Analyze file metadata/content using Gemini.
        
        Args:
            file_data (str): String representation of file details (e.g., list of files)
        
        Returns:
            str: Gemini-generated analysis with file types, recommendations, and next steps
        """
        try:
            # Define prompt to instruct Gemini as a Data Analysis Agent
            prompt = f"""
            You are a Data Analysis Agent. Analyze these files:

            {file_data}

            Provide:
            1. File types found
            2. Recommendations
            3. Next steps

            Keep it short and actionable.
            """

            # Call Gemini model to generate analysis
            response = self.model.generate_content(prompt)

            # Return the text content of the response
            return response.text

        except Exception as e:
            # Handle errors gracefully and return error message
            return f"Analysis error: {str(e)}"


# -------------------------
# Create an instance of Analysis Agent
# -------------------------
analysis_agent = AnalysisAgent()
print("✅ Analysis Agent created with Gemini 1.5 Flash")
```

## Cell 6: Define Report Agent

```python
class ReportAgent:
    """Report Agent using Gemini 1.5 Flash"""

    def __init__(self):
        # Configure Gemini API with provided API key
        genai.configure(api_key=GEMINI_API_KEY)
        # Initialize Gemini Generative Model (Flash for fast, cost-efficient responses)
        self.model = genai.GenerativeModel('gemini-1.5-flash')

    def create_report(self, data):
        """
        Generate a professional report using Gemini.

        Args:
            data (str): Input data/content to summarize (e.g., analysis results, raw text)

        Returns:
            str: Generated report text containing summary, findings, and recommended actions
        """
        try:
            # Define prompt instructing Gemini to act as a "Report Agent"
            prompt = f"""
            You are a Report Agent. Create a brief summary:

            {data}

            Provide:
            1. Summary
            2. Key findings
            3. Recommended actions

            Keep it concise and professional.
            """

            # Call Gemini to generate structured report
            response = self.model.generate_content(prompt)

            # Extract and return the response text
            return response.text

        except Exception as e:
            # Handle API or runtime errors gracefully
            return f"Report error: {str(e)}"


# -------------------------
# Create instance of Report Agent
# -------------------------
report_agent = ReportAgent()
print("✅ Report Agent created with Gemini 1.5 Flash")
```

## Cell 7: Define Multi-Agent Coordinator

```python
class MultiAgentCoordinator:
    """Coordinates all agents (GCP, Analysis, Report) using Semantic Kernel with Gemini"""

    def __init__(self, kernel):
        # Store reference to the Semantic Kernel (SK) instance
        self.kernel = kernel

    async def run_workflow(self):
        """
        Orchestrates the full multi-agent workflow:
        1. GCP Agent fetches files
        2. Analysis Agent analyzes files
        3. Report Agent generates report
        """
        print("🚀 Multi-Agent Workflow Starting...")

        # -------------------------
        # Step 1: Get files from GCP Agent via Semantic Kernel
        # -------------------------
        print("\n📋 GCP Agent: Getting files...")
        files = await self.kernel.invoke(self.kernel.plugins["gcp"]["list_files"])
        print(f"GCP Result: {files}")

        # -------------------------
        # Step 2: Analyze files with Gemini-powered Analysis Agent
        # -------------------------
        print("\n🧠 Analysis Agent: Analyzing...")
        analysis = analysis_agent.analyze_files(str(files))
        # Print preview (first 100 chars only to avoid clutter)
        print(f"Analysis: {analysis[:100]}...")

        # -------------------------
        # Step 3: Generate report using Gemini-powered Report Agent
        # -------------------------
        print("\n📊 Report Agent: Creating report...")
        report = report_agent.create_report(f"Files: {files}\nAnalysis: {analysis}")
        # Print preview (first 100 chars)
        print(f"Report: {report[:100]}...")

        # Return all results for downstream use
        return {
            "files": str(files),
            "analysis": analysis,
            "report": report
        }

    async def upload_and_analyze(self, filename):
        """
        Upload a file and run analysis:
        1. Check local files
        2. Upload selected file to GCP bucket
        3. List files in bucket
        4. Analyze bucket contents with Analysis Agent
        """
        print(f"📤 Upload and Analyze: {filename}")

        # -------------------------
        # Step 1: Check local files before uploading
        # -------------------------
        local_check = await self.kernel.invoke(self.kernel.plugins["gcp"]["check_local_files"])
        print(f"Local files check: {local_check}")

        # -------------------------
        # Step 2: Upload file to GCP bucket via Semantic Kernel
        # -------------------------
        upload_result = await self.kernel.invoke(
            self.kernel.plugins["gcp"]["upload_file"],
            filename=filename
        )
        print(f"Upload: {upload_result}")

        # -------------------------
        # Step 3: Get updated file list from GCP bucket
        # -------------------------
        files = await self.kernel.invoke(self.kernel.plugins["gcp"]["list_files"])

        # -------------------------
        # Step 4: Analyze files with Analysis Agent
        # -------------------------
        analysis = analysis_agent.analyze_files(str(files))

        # Return combined result
        return {"upload": str(upload_result), "analysis": analysis}


# -------------------------
# Create Multi-Agent Coordinator instance
# -------------------------
coordinator = MultiAgentCoordinator(kernel)
print("✅ Multi-Agent Coordinator ready")
```

## Cell 8: Define Quick Commands

```python
# -------------------------
# Quick utility functions for Semantic Kernel + Gemini agents
# -------------------------

async def quick_list():
    """Quickly list files in GCP bucket & local directory using Semantic Kernel"""
    # Call GCP Agent's list_files function via kernel
    files = await kernel.invoke(kernel.plugins["gcp"]["list_files"])
    print(files)  # Print results for visibility
    return files  # Return results to caller

async def quick_check_local():
    """Quickly check local files using Semantic Kernel"""
    # Call GCP Agent's check_local_files function via kernel
    files = await kernel.invoke(kernel.plugins["gcp"]["check_local_files"])
    print(files)  # Print results
    return files

async def quick_upload(filename):
    """Quickly upload a file to GCP bucket using Semantic Kernel"""
    # Call GCP Agent's upload_file function via kernel
    result = await kernel.invoke(kernel.plugins["gcp"]["upload_file"], filename=filename)
    print(result)  # Print upload status
    return result

async def quick_analyze():
    """Quickly analyze files using Gemini Analysis Agent"""
    # Step 1: Get current file list from GCP
    files = await kernel.invoke(kernel.plugins["gcp"]["list_files"])
    # Step 2: Run Gemini-powered Analysis Agent on file list
    analysis = analysis_agent.analyze_files(str(files))
    print(analysis)  # Print analysis results
    return analysis

async def quick_workflow():
    """Run the full multi-agent workflow (GCP → Analysis → Report)"""
    return await coordinator.run_workflow()

async def quick_upload_and_analyze(filename):
    """Upload a file and then analyze files using multi-agent flow"""
    return await coordinator.upload_and_analyze(filename)


# -------------------------
# Usage guide (helper prints)
# -------------------------
print("✅ Enhanced quick commands ready:")
print("- await quick_check_local()          # Check local files")
print("- await quick_list()                 # List bucket & local files")
print("- await quick_upload('filename')     # Smart upload")
print("- await quick_analyze()              # Multi-agent analysis with Gemini")
print("- await quick_workflow()             # Full workflow (files → analysis → report)")
print("- await quick_upload_and_analyze('filename')  # Upload + analyze in one go")
```
