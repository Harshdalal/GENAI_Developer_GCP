#✅ Step 3: Initialize the LLM

```python
from llama_index.llms.google_genai import GoogleGenAI

llm = GoogleGenAI(model="models/gemini-1.5-flash-latest", temperature=0.3)
```

#✅ Step 4: Define Multi-Agent A2A Protocol Communication

```python
from llama_index.core.llms import ChatMessage, MessageRole

# Agent A (Domain Expert) initiates discussion
agent_a = ChatMessage(
    role=MessageRole.SYSTEM,
    content="You're a domain expert in Generative AI guiding research direction."
)

# Agent B (Experimental Analyst) asks for review
agent_b = ChatMessage(
    role=MessageRole.USER,
    content="Can you outline the key breakthroughs in Generative AI over the past 5 years?"
)

# Agent C (Data Modeler) adds insights
agent_c = ChatMessage(
    role=MessageRole.ASSISTANT,
    content="Sure. Recent breakthroughs include transformer architectures, diffusion models, prompt engineering, and fine-tuning strategies like LoRA."
)

# Combine agent dialogue as message history
conversation = [agent_a, agent_b, agent_c]
```

#✅ Step 4A: Simulate Agent Collaboration

```python
# Let the LLM respond to entire conversation with new goal
response = llm.chat(conversation)
print("\n🔁 Final Agent Collaboration Output:\n")
print(response)
```

#✅ Step 5: Add Model Context Protocol (MCP)

```python
# New context added (like a real-time goal change)
mcp_context = ChatMessage(
    role=MessageRole.USER,
    content="Can you provide a comparative analysis with figures or tabular summary?"
)

# Append context to message thread
conversation.append(mcp_context)
```

#✅ Step 6: Simulate Agent Collaboration

```python
# Let the LLM respond to entire conversation with new goal
response = llm.chat(conversation)
print("\n🔁 Final Agent Collaboration Output:\n")
print(response)
```

#✅ Step 7: (Optional) Stream LLM Output as if Real-Time Agent Collaboration

```python
stream = llm.stream_chat(conversation)
print("\n🔄 Streaming Collaboration:\n")
for chunk in stream:
    print(chunk.delta, end="")
```
