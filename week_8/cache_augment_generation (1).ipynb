#🧠 Step 2: Define Simple Cache + Augment Function

```python
# Simulated cache (could be Redis or vector DB in advanced version)
cache = {}

def cache_augment_generation(user_prompt):
    # Check if the prompt or similar content exists in cache
    matched_contexts = [v for k, v in cache.items() if k in user_prompt]
    context = "\n".join(matched_contexts[-2:])  # Get last 2 related cached generations

    # Augment user prompt with retrieved context
    full_prompt = f"{context}\n\nUser Prompt: {user_prompt}".strip()

    # Generate using Gemini
    response = model.generate_content(full_prompt)

    # Cache the result
    cache[user_prompt] = response.text

    return response.text
```

#▶️ Step 3: Test the Cache-Augmented Generation

```python
# First time: generate with no cache
response1 = cache_augment_generation("Explain Newton's First Law.")
print("Response 1:\n", response1)

# Second time: new query but overlaps with earlier topic
response2 = cache_augment_generation("How is inertia related to Newton's First Law?")
print("\nResponse 2:\n", response2)

# Third: another related follow-up that uses cached context
response3 = cache_augment_generation("Give a real-world example of Newton's First Law.")
print("\nResponse 3:\n", response3)
```
